````r
library(readr)
library(dplyr)
library(ggplot2)
library(tidyverse)
````

````r
data <- read.csv("UsedCars.csv")
summary(data)
head(data, 10)
````

````r
#Run a linear regression of Price on all available variables
model <- lm(Price ~ Age + KM + HP + Metallic + Automatic + CC + Doors + Gears + Weight, data = data)
summary(model)
````

It seems like weight, Gears, whether a car is automatic, HP, KM, Age are affecting the car price. These variables have stronger influence with statistical significance. Overall model p value<0.05, therefore doesnt reject the null hypothesis and is statistical significant. The current list of variables could be used to explain 86% of the model. Automatic has the lagest coefficient estimate.



````r
# Calculate fitted values and residuals
fitted_values <- fitted(model)  # or predict(model)
residuals_values <- residuals(model)  # or model$residuals

# Create a comparison table
results_table <- data.frame(
  Original_Y = data$Price,           # Original y values
  Fitted_Y = fitted_values,          # Fitted ŷ values
  Residuals = residuals_values       # Residuals (y - ŷ)
)

# Display the table
print(results_table)

# Or view first 20 rows for better readability
head(results_table, 20)
````

````r
#Calculate t-statstics
t_stats <- summary(model)$coefficients[, "t value"]
print(t_stats)
````

````r
# Step 1: Get the degrees of freedom from the regression model
df <- df.residual(model)
print(paste("Degrees of freedom:", df))


# Step 2: Calculate the critical value for 95% confidence level
# For a two-tailed test at 95% confidence, we need the 97.5th percentile
# (because we have 2.5% in each tail: 2.5% + 95% + 2.5% = 100%)
critical_value <- qt(0.975, df = df)
print(paste("Critical value at 95% confidence level:", critical_value))

# Alternative way to see both positive and negative critical values
cat("Critical values: ±", critical_value, "\n")
````

````r

# Calculate VIF for all variables
vif_values <- vif(model)
print(vif_values)

# Extract VIF specifically for KM variable
vif_km <- vif_values["KM"]
print(paste("VIF for KM variable:", vif_km))
````

Interpreting VIF for multicollinearity:
VIF interpretation guidelines:

VIF = 1: No multicollinearity

1 < VIF < 5: Moderate multicollinearity (generally acceptable)

5 ≤ VIF < 10: High multicollinearity (concerning)

VIF ≥ 10: Severe multicollinearity (problematic)


What VIF means:

VIF = 5 means the variance of the coefficient is 5 times larger due to multicollinearity
VIF = 10 means the standard error is √10 ≈ 3.16 times larger than it would be without multicollinearity

````r
# Regress Weight on all other explanatory variables (excluding Price)
weight_model <- lm(Weight ~ Age + KM + HP + Metallic + Automatic + CC + Doors + Gears, data = data)
summary(weight_model)
````

````r

# Get the R-squared value
r_squared_weight <- summary(weight_model)$r.squared
print(paste("R-squared for Weight regression:", r_squared_weight))
vif_weight <- 1 / (1 - r_squared_weight)
print(paste("VIF for Weight:", vif_weight))
````

It shows how well the other variables predict Weight:

R² close to 0 → Weight is not well predicted by other variables → low multicollinearity

R² close to 1 → Weight is highly predicted by other variables → high multicollinearity

````r
data2 <- read.csv("Loan.csv")

summary(data2)
head(data2, 10)
##CCAvg= Average spending on credit cards per month in $1000
##Income = Avg income of customers in $1000
##Loan = whether or not the customer accepted hte persoanl loan offer 1- yes 0-no
##Education level: 1: Undergrad; 2: Graduate; 3: Advanced/Professional
##Family = Family size of the customer
````

````r
data2$Education <- as.factor(data2$Education)
````

````r
# Regress Loan (binary outcome) on all other variables
lpm_model <- lm(Loan ~ Income + Family + CCAvg + Education, data = data2)

# Show the summary of regression results
summary(lpm_model)
````

````r
# Get fitted values from the linear probability model
fitted_values <- fitted(lpm_model)  # or predict(lmp_model)

# Find customers with fitted values > 1 or < 0
customers_over_1 <- which(fitted_values > 1)
customers_under_0 <- which(fitted_values < 0)

print(customers_over_1)
length(customers_over_1)
length(customers_under_0)
````

````r
# Get fitted values from the linear probability model
fitted_values <- fitted(lpm_model)  # Make sure you're using the correct model name

# Find the smallest fitted value
min_fitted <- min(fitted_values)
print(paste("Smallest fitted y value:", min_fitted))
````

````r
# Run Logit model using glm() function
logit_model <- glm(Loan ~ Income + Family + CCAvg + Education,
                   data = data2,
                   family = binomial(link = "logit"))

# Show the summary of estimation results
summary(logit_model)
````

````r
# Get fitted probabilities from logit model
fitted_probabilities <- predict(logit_model, type="response")

# Calculate threshold (fraction of success in original data)
threshold <- mean(data2$Loan)

# Create binary predictions
predicted_loan <- ifelse(fitted_probabilities > threshold, 1, 0)

# Create confusion matrix
confusion_matrix <- table(Actual = data2$Loan, Predicted = predicted_loan)
print(confusion_matrix)

# Calculate overall PCP
total_correct <- sum(diag(confusion_matrix))
total_observations <- sum(confusion_matrix)
overall_pcp <- (total_correct / total_observations) * 100

cat("Overall PCP:", round(overall_pcp, 2), "%")
````

````r
# prompt: what are the correct PCP values for y=1 and y=0, respectively?

# Correctly Classified for y=1 (Actual 1, Predicted 1)
pcp_y1 <- confusion_matrix[2, 2] / sum(confusion_matrix[2, ]) * 100

# Correctly Classified for y=0 (Actual 0, Predicted 0)
pcp_y0 <- confusion_matrix[1, 1] / sum(confusion_matrix[1, ]) * 100

cat("\nPCP for y=1:", round(pcp_y1, 2), "%")
cat("\nPCP for y=0:", round(pcp_y0, 2), "%")
````

````r
# prompt: Calculate the predicted probability of success according to the Logit model estimation
# results. Evaluate the probability at such values of the X variables: {Income, Family, CCAvg}
# equal their mean values in the original data and Education=”2”.
# • First calculate the probability using the defining formula as explained in class. You may
# do the calculation in R, but be explicit about the exact formula used and substitute in
# the exact numbers.
# • Then use the predict(…, type=”response”) function to obtain the predicted probability
# of success. The value should be the same as your own calculation result above.

# Define the values for the calculation
income_mean <- mean(data2$Income)
family_mean <- mean(data2$Family)
ccavg_mean <- mean(data2$CCAvg)
education_level <- "2" # Corresponds to the factor level "2"

# Get the coefficients from the logit model
coefficients <- coef(logit_model)
intercept <- coefficients["(Intercept)"]
income_coef <- coefficients["Income"]
family_coef <- coefficients["Family"]
ccavg_coef <- coefficients["CCAvg"]
education_coef <- coefficients[paste0("Education", education_level)] # Get coefficient for the specific education level

# Calculate the linear predictor (Z) using the formula
Z <- intercept + (income_mean * income_coef) + (family_mean * family_coef) + (ccavg_mean * ccavg_coef) + education_coef

# Calculate the predicted probability using the logistic function: P = 1 / (1 + exp(-Z))
predicted_prob_manual <- 1 / (1 + exp(-Z))

cat("Manual Calculation of Predicted Probability:\n")
cat("  Formula: P = 1 / (1 + exp(-Z))\n")
cat("  Where Z = (Intercept) + (Income * Coef_Income) + (Family * Coef_Family) + (CCAvg * Coef_CCAvg) + (Education_2 * Coef_Education_2)\n")
cat("  Substituting values:\n")
cat("    Z = ", intercept, " + (", income_mean, " * ", income_coef, ") + (", family_mean, " * ", family_coef, ") + (", ccavg_mean, " * ", ccavg_coef, ") + (1 * ", education_coef, ")\n", sep="")
cat("    Z = ", Z, "\n")
cat("  Predicted Probability = 1 / (1 + exp(-", Z, ")) = ", predicted_prob_manual, "\n\n", sep="")


# Use the predict function to obtain the predicted probability
# Create a new data frame with the specific values for prediction
newdata <- data.frame(
  Income = income_mean,
  Family = family_mean,
  CCAvg = ccavg_mean,
  Education = as.factor(education_level)
)

predicted_prob_predict <- predict(logit_model, newdata = newdata, type = "response")

cat("Predicted Probability using predict() function:\n")
cat("  Predicted Probability = ", predicted_prob_predict, "\n")

# Verify if the results are the same (within a small tolerance for floating point errors)
tolerance <- 1e-6
if (abs(predicted_prob_manual - predicted_prob_predict) < tolerance) {
  cat("\nManual calculation and predict() function results match.\n")
} else {
  cat("\nWarning: Manual calculation and predict() function results do not exactly match.\n")
}
````

````r
# prompt: Calculate the partial effects of all independent variables based on the Logit model. Again,
# evaluate the probability at such values of the independent variables: {Income, Family,
# CCAvg} equal their mean values in the original data and Education=”2”. Also calculate the
# partial effects based on the linear probability model (Hint: simply the 𝛽𝛽 coefficients). Co-list
# and compare the partial effects from the two models. Are they comparable

# --- Logit Model Partial Effects ---

# Calculate the predicted probability at the specified values (already done above)
# predicted_prob_at_means_edu2 <- predicted_prob_predict # This is the probability P(Y=1 | X=x_bar, Edu=2)

# Get the coefficients from the Logit model
logit_coefs <- coef(logit_model)

# Calculate partial effects for continuous variables (Income, Family, CCAvg)
# PE = beta_k * P(Y=1 | X) * (1 - P(Y=1 | X))
pe_income_logit <- logit_coefs["Income"] * predicted_prob_predict * (1 - predicted_prob_predict)
pe_family_logit <- logit_coefs["Family"] * predicted_prob_predict * (1 - predicted_prob_predict)
pe_ccavg_logit <- logit_coefs["CCAvg"] * predicted_prob_predict * (1 - predicted_prob_predict)

# Calculate partial effect for dummy variable (Education, assuming baseline is Education=1)
# The PE for moving from Education=1 to Education=2 (at the specified values)
# is P(Y=1 | Income, Family, CCAvg, Edu=2) - P(Y=1 | Income, Family, CCAvg, Edu=1)

# Probability at Edu=2 (already calculated as predicted_prob_predict)
prob_edu2 <- predicted_prob_predict

# Probability at Edu=1 (keeping other variables at specified means)
newdata_edu1 <- data.frame(
  Income = income_mean,
  Family = family_mean,
  CCAvg = ccavg_mean,
  Education = as.factor("1") # Set Education to level 1
)
prob_edu1 <- predict(logit_model, newdata = newdata_edu1, type = "response")

# Partial effect for Education (moving from 1 to 2)
pe_edu_logit <- prob_edu2 - prob_edu1

cat("\n--- Logit Model Partial Effects at Mean Income/Family/CCAvg and Education='2' ---\n")
cat("Partial Effect of Income:  ", pe_income_logit, "\n")
cat("Partial Effect of Family:  ", pe_family_logit, "\n")
cat("Partial Effect of CCAvg:   ", pe_ccavg_logit, "\n")
cat("Partial Effect of Education (1->2): ", pe_edu_logit, "\n")


# --- Linear Probability Model Partial Effects ---

# Get the coefficients from the LPM model
lpm_coefs <- coef(lpm_model)

# Partial effects for LPM are simply the coefficients
pe_income_lpm <- lpm_coefs["Income"]
pe_family_lpm <- lpm_coefs["Family"]
pe_ccavg_lpm <- lpm_coefs["CCAvg"]
# For the dummy variable Education, the coefficient represents the change
# in probability when moving from the base level (Education=1) to the specified level (Education=2).
pe_edu_lpm <- lpm_coefs["Education2"]

cat("\n--- Linear Probability Model Partial Effects (Coefficients) ---\n")
cat("Partial Effect of Income:  ", pe_income_lpm, "\n")
cat("Partial Effect of Family:  ", pe_family_lpm, "\n")
cat("Partial Effect of CCAvg:   ", pe_ccavg_lpm, "\n")
cat("Partial Effect of Education (1->2): ", pe_edu_lpm, "\n")


# --- Comparison ---

cat("\n--- Comparison of Partial Effects (Logit vs LPM) ---\n")
comparison_table <- data.frame(
  Variable = c("Income", "Family", "CCAvg", "Education (1->2)"),
  Logit_PE = c(pe_income_logit, pe_family_logit, pe_ccavg_logit, pe_edu_logit),
  LPM_PE = c(pe_income_lpm, pe_family_lpm, pe_ccavg_lpm, pe_edu_lpm)
)
print(comparison_table)

cat("\nAre the partial effects comparable?\n")
cat("Generally, yes, the signs and relative magnitudes should be broadly similar.\n")
cat("However, the *exact* values will differ because:\n")
cat("1. LPM assumes a linear relationship between predictors and probability,\n")
cat("   which is incorrect for a binary outcome (probabilities must be between 0 and 1).\n")
cat("2. Logit accounts for the non-linear S-shaped relationship of probability.\n")
cat("   Logit partial effects for continuous variables depend on the specific values of the predictors (P(Y=1|X) term).\n")
cat("In this case, let's look at the table:\n")
# Check the differences
diff_income <- abs(pe_income_logit - pe_income_lpm)
diff_family <- abs(pe_family_logit - pe_family_lpm)
diff_ccavg <- abs(pe_ccavg_logit - pe_ccavg_lpm)
diff_edu <- abs(pe_edu_logit - pe_edu_lpm)

cat(paste0("- Income PE difference: ", round(diff_income, 4), "\n"))
cat(paste0("- Family PE difference: ", round(diff_family, 4), "\n"))
cat(paste0("- CCAvg PE difference: ", round(diff_ccavg, 4), "\n"))
cat(paste0("- Education PE difference: ", round(diff_edu, 4), "\n"))

cat("The differences suggest that the LPM likely over/underestimates the effects compared to the Logit model, depending on the variable and the point of evaluation.\n")
cat("LPM is simpler but can produce probabilities outside [0,1] and has constant partial effects (the coefficients).\n")
cat("Logit is more theoretically sound for binary outcomes and has partial effects that vary with the predictor values.\n")
cat("Based on these differences, while broadly aligned, the Logit model provides a more accurate estimate of the change in probability for a unit change in a predictor at this specific point in the data (means + Edu=2).\n")
cat("For continuous variables, the Logit PE is the slope of the probability curve at that point, while the LPM PE is the average slope assuming linearity.\n")
cat("For dummy variables, the Logit PE is the discrete change in probability, while the LPM PE is simply the coefficient difference.\n")
cat("The comparison table shows that the numerical values are different, highlighting the non-linearity captured by the Logit model compared to the linear approximation of the LPM.\n")
````

````r
# prompt: "They are not directly comparable because the partial effects from the linear probability model equal the coefficients, but the partial effects from the logit model are transformations of the coefficients." is this statement true

# Yes, that statement is true.
# The partial effects in a Linear Probability Model (LPM) are indeed equal to the coefficients of the explanatory variables. A unit change in a predictor directly changes the predicted probability by its coefficient value.
# In contrast, the partial effects (or marginal effects) in a Logit model are *not* simply the coefficients. They are calculated based on the logistic function's derivative (for continuous variables) or as the difference in predicted probabilities (for discrete/dummy variables) at a specific point in the data (e.g., at the mean of the predictors). The formula for the partial effect of a continuous variable X_k in a Logit model is beta_k * Lambda(beta'X), where beta_k is the coefficient for X_k, and Lambda(.) is the PDF of the logistic distribution evaluated at the linear predictor beta'X.
# This non-linear transformation means the partial effects in a Logit model vary depending on the values of *all* the predictors, while they are constant in an LPM. Therefore, the numerical values of partial effects are generally different between the two models, making the statement accurate.
````
